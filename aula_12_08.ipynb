```python
#exercico 06

#considere:
notas = [7.5, 8.0, 6.5, 9.0, 5.5]
#mostre:
#a maior nota
#a menor nota
#a media das notas

maior = max(notas)
print(f"A maior nota é: {maior}")

menor = min(notas)
print(f"A menor nota é: {menor}")

media = sum(notas) / len(notas)
print(f"A média das notas é: {media}")
```

```python
#exercicio 07

# crie uma lista vazia chamada produtos
# peca ao usuario para cadastrar 5 produtos utilizando for
#mostre os produtos numerados

# exp:
# 1 - arroz
# 2 - feijao
# 3 - leite

produtos = []
for _ in range(5):
  produto_digitado = input("Digite um produto: ")
  produtos.append(produto_digitado)
for numero, produto in enumerate(produtos, start=1):
    print(f"{numero} - {produto}")
```

```python
# exercicio 08

#considere:
alunos = ["ana", "carlos", "maria", "pedro", "joao"]

#peca ao usuario para digitar um nome, o programa devera verificar se o nome consta na lista ou nao
#faca o programa funcionar independente da forma que o usuario digitar.
#exp: Marie, maria, MARIA


print("Busca de Alunos")
busca = input("Digite o nome do aluno que deseja procurar: ")

busca_formatada = busca.strip().lower()

if busca_formatada in alunos:
    print(f"Sucesso! O(A) aluno(a) '{busca_formatada.capitalize()}' está na lista.")
else:
    print(f"Aviso: O nome '{busca}' não foi encontrado na lista de alunos.")
```

```python
# exercicio 09

# crie uma lista vazia chamada numeros
#peca ao usuario para informar 10 numeros
#depois crie duas listas chamadas pares e impares
#percorra a lista original e coloque cada numero na lista correspondente

numeros = []
pares = []
impares = []

for i in range(10):
    n = int(input(f"Informe o {i+1}º número: "))
    numeros.append(n)

for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)

print("")
print("Pares:", pares)
print("Ímpares:", impares)

```

```python
#exercicio 10

#faca um pequeno programa sistema para cadastrar as notas de 05 alunos.
#o programa devera armazenar as notas em uma lista e depois mostrar:

#todas as notas
#maior nota
#media da turma
#quantidade de alunos com nota maior ou igual a 6
#quantidade de alunos com nota menor ou igual a 6

notas = []

print("Cadastro de Notas")
for i in range(5):
    while True:
        try:
            nota = float(input(f"Digite a nota do aluno {i + 1} (0 a 10): "))
            if 0 <= nota <= 10:
                notas.append(nota)
                break
            else:
                print("Por favor, digite uma nota válida entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Digite apenas números.")


maior_nota = max(notas)
media_turma = sum(notas) / len(notas)

maior_igual_seis = sum(1 for nota in notas if nota >= 6)
menor_igual_seis = sum(1 for nota in notas if nota <= 6)

print("")

print("Relatório Final")
print(f"Todas as notas cadastradas: {notas}")
print(f"Maior nota da turma: {maior_nota:.1f}")
print(f"Média da turma: {media_turma:.1f}")
print(f"Quantidade de alunos com nota >= 6: {maior_igual_seis}")
print(f"Quantidade de alunos com nota <= 6: {menor_igual_seis}")
```
