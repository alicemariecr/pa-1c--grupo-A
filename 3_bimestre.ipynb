```python
#Escrever um programa que indentifica se um numero é positivo ou negativo
numero = int(input("Digite um número: "))
if numero >=0:
  print("Esse número é positivo.")
else:
  print("Esse número é negativo.")
```

```python
#Escrever um programa que recebe um número que 20 e imprime os números pares
numero = int(input("Digite um número menor que 20: "))
num = numero + 1
if numero >= 20 or numero <=0:
  print("Número inválido.")
else:
  for i in range (0, num, 2):
    print(i)
```

```python
#Escrever um programa q ue imprime numeros de 0 a 5 usando while
contador = 0
while contador <= 5:
  print(contador)
  contador += 1
```

```python
#Escrever um programa que recebe um número que 20 e imprime os números pares
numb = int(input("Digite um número menor que 20: "))
num = 0
if numb > 20 or numb < 0:
  print("Número inválido.")
while num <= numb and numb <= 20:
  if num % 2 == 0:
    print(num)
  num += 1
```
