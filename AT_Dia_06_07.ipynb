```python
#EXERCÍCIO 3:
#COFRINHO

meta = 10000
total = 0
meses = 0

deposito = float(input("Digite o valor que será depositado por mês: R$ "))

while total < meta:
    total += deposito
    meses += 1

print("Meses necessários:", meses)
print("Total acumulado: R$", total)
print("A meta foi atingida no mês", meses)
```

```python
#EXERCÍCIO 4:
#CAIXA REGISTRADORA

subtotal = 0

while True:
    preco = float(input("Preço do produto (CLIQUE 0 para finalizar): "))

    if preco == 0:
        break

    quantidade = int(input("Quantidade: "))

    subtotal += preco * quantidade

if subtotal > 1000:
    desconto = 0.15
elif subtotal > 500:
    desconto = 0.10
elif subtotal > 100:
    desconto = 0.05
else:
    desconto = 0

valor_desconto = subtotal * desconto
valor_final = subtotal - valor_desconto

print("\nSubtotal: R$", subtotal)
print("Desconto:", desconto * 100, "%")
print("Valor do desconto: R$", valor_desconto)
print("Valor final: R$", valor_final)
```
