```python
#10

height = int(input("Enter the height of the pyramid: "))

for i in range(1, height + 1):
    # Build the list of number strings first, then concatenate manually
    temp_numbers = []
    for j in range(i, 0, -1): # Decreasing sequence
        temp_numbers.append(str(j))
    for j in range(2, i + 1): # Increasing sequence (starts from 2)
        temp_numbers.append(str(j))

    # Manually concatenate the numbers with spaces, without using .join()
    numbers_str = ""
    for idx, num_str in enumerate(temp_numbers):
        numbers_str += num_str
        if idx < len(temp_numbers) - 1:
            numbers_str += " "

    # Calculate leading spaces for centering. Each number and its space take roughly 2 characters.
    leading_spaces = " " * (2 * (height - i))

    print(leading_spaces + numbers_str)
```

```python
#media com aprovado reprovado e recuperação

n1 = int(input("Digite a nota 1: "))
n2 = int(input("Digite a nota 2: "))
n3 = int(input("Digite a nota 3: "))
n4 = int(input("Digite a nota 4: "))

media = (n1 + n2 + n3 + n4) / 4

if media >= 7:
  print(f"Prabens você tirou {media} na media, APROVADO")
elif media >= 5 and media < 7:
  print(f"Você tirou {media} na media, RECUPERAÇÃO")
else:
  print(f"Você tirou {media} na media, REPROVADO")
```

```python
# 8 faça um programa que recebe um valores e quantidades de produto e no final  exiba o valor da compra
total = 0
while True:
    produto = input("Digite o nome do produto (ou 'sair' para encerrar): ")
    if produto.lower() == 'sair':
        break
    preco = float(input("Digite o preço do produto: "))
    quantidade = int(input("Digite a quantidade comprada: "))
    total += preco * quantidade
print(f"O valor total da compra é: R${total:.2f}")
```
